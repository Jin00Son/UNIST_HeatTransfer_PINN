# PyTorch PINN Ablation Runner

Current target: Stage 5 noise-seed ablation.

This suite fixes the final two-row sensor placement, sigma_theta = 0.002, theta_y_data = 1, and smooth = 0. It changes only the random sensor-noise seed to show how the spatial pattern of independent sensor errors affects inverse Bi recovery.

```python
from pathlib import Path
import os
import subprocess
import sys
import time

PROJECT_DIR = Path.cwd()
if not (PROJECT_DIR / "PINN_pytorch_ablation_runner.py").exists():
    PROJECT_DIR = Path("/home/jinooson/DL_study/Heat_Transfer_PINN/Final_Project")

RUNNER = PROJECT_DIR / "PINN_pytorch_ablation_runner.py"
assert RUNNER.exists(), RUNNER

print("project dir:", PROJECT_DIR)
print("runner     :", RUNNER)
```

## Run Configuration

Set `N_ADAM_OVERRIDE` only for quick trial runs. Leave it as `None` for the full ablation settings in the runner.

```python
SUITES = [
    "sensor_stage5_noise_seed_fd",
]

# Full study: keep these defaults.
DEVICE = "cuda"  # use "auto" to fall back to CPU, or "cuda:1" for a specific GPU
N_ADAM_OVERRIDE = None
SKIP_LBFGS = False
MAX_RUNS = None

print("Suites to run:", SUITES)
```

```python
def run_suite(suite):
    cmd = [sys.executable, "-u", str(RUNNER), "--suite", suite, "--device", DEVICE]
    if N_ADAM_OVERRIDE is not None:
        cmd += ["--n-adam", str(N_ADAM_OVERRIDE)]
    if SKIP_LBFGS:
        cmd += ["--no-lbfgs"]

    env = os.environ.copy()
    env.setdefault("MPLCONFIGDIR", "/tmp/matplotlib")
    env.setdefault("PYTHONUNBUFFERED", "1")

    print("\n" + "=" * 90)
    print(f"START SUITE: {suite}")
    print("command:", " ".join(cmd))
    t0 = time.perf_counter()
    subprocess.run(cmd, cwd=PROJECT_DIR, env=env, check=True)
    print(f"END SUITE: {suite} | elapsed={time.perf_counter() - t0:.1f}s")
```

## Run All Selected Suites

Executing this cell starts the selected ablation suites sequentially.

```python
for suite in SUITES:
    run_suite(suite)

print("\nAll selected ablation suites finished.")
```
